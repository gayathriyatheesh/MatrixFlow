```python
data = pd.read_csv("../ctr_features.csv")

print(data.shape)

data.head()
```

```python
data["clicked"] = np.where(
    data["rating"] >= 4,
    1,
    0
)

data.head()
```

```python
print(data["clicked"].value_counts())
```

```python
features = data[
    [
        "avg_rating",
        "rating_count",
        "avg_movie_rating",
        "movie_rating_count"
    ]
]

target = data["clicked"]

print(features.head())
print(target.head())
```

```python
print(features.shape)
print(target.shape)
```

```python
features.to_csv("../ctr_X.csv", index=False)

target.to_csv("../ctr_y.csv", index=False)

print("Training dataset saved successfully!")
```

```python
positive_samples = data[data["rating"] >= 4].copy()

positive_samples["label"] = 1

print("Positive samples:", positive_samples.shape)
```

```python
negative_samples = data[data["rating"] < 4].copy()

negative_samples["label"] = 0

print("Negative samples:", negative_samples.shape)
```

```python
n = min(len(positive_samples), len(negative_samples))

positive_samples = positive_samples.sample(
    n=n,
    random_state=42
)

negative_samples = negative_samples.sample(
    n=n,
    random_state=42
)

print(len(positive_samples))
print(len(negative_samples))
```

```python
ctr_dataset = pd.concat(
    [positive_samples, negative_samples],
    ignore_index=True
)

ctr_dataset = ctr_dataset.sample(
    frac=1,
    random_state=42
).reset_index(drop=True)

print(ctr_dataset.shape)
```

```python
X = ctr_dataset[
    [
        "avg_rating",
        "rating_count",
        "avg_movie_rating",
        "movie_rating_count"
    ]
]

y = ctr_dataset["label"]

print(X.shape)
print(y.shape)
```

```python
X.to_csv("../ctr_X.csv", index=False)

y.to_csv("../ctr_y.csv", index=False)

print("Final CTR training dataset saved!")
```
