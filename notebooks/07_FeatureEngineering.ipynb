```python
import pandas as pd
import numpy as np
```

```python
ratings = pd.read_csv("../ml-25m/ratings.csv")

print(ratings.shape)

ratings.head()
```

```python
ratings = ratings.sample(
    n=500000,
    random_state=42
).reset_index(drop=True)

print(ratings.shape)
```

```python
user_features = ratings.groupby("userId").agg(
    avg_rating=("rating", "mean"),
    rating_count=("rating", "count")
).reset_index()

user_features.head()
```

```python
movie_features = ratings.groupby("movieId").agg(
    avg_movie_rating=("rating", "mean"),
    movie_rating_count=("rating", "count")
).reset_index()

movie_features.head()
```

```python
data = ratings.merge(
    user_features,
    on="userId"
)

data.head()
```

```python
data = data.merge(
    movie_features,
    on="movieId"
)

data.head()
```

```python
print(data.columns)
```

```python
data.to_csv("../ctr_features.csv", index=False)

print("CTR features saved successfully!")
```
